```python
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
import numpy as np

from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
digits = load_iris()
```

```python
dir(digits)
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(digits.data,digits.target,test_size=0.3)
```

```python
from sklearn.model_selection import KFold
kf = KFold(n_splits=3)
kf
```

```python
def get_score(model,X_train, X_test, y_train, y_test):
    model.fit(X_train,y_train)
    model.score(X_test,y_test)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
l_scores = cross_val_score(LogisticRegression(), digits.data, digits.target)
l_scores
```

```python
np.average(l_scores)
```

```python
from sklearn.tree import DecisionTreeClassifier
```

```python
d_score=cross_val_score(DecisionTreeClassifier(), digits.data, digits.target,cv=3)
```

```python
np.average(d_score)
```

```python
s_scores = cross_val_score(SVC(), digits.data, digits.target)
s_scores
```

```python
cross_val_score(SVC(), digits.data, digits.target).mean()
```

```python
np.average(s_scores)
```

```python
r_scores = cross_val_score(RandomForestClassifier(n_estimators=50), digits.data,digits.target)
r_scores
```

```python
np.average(r_scores)
```
